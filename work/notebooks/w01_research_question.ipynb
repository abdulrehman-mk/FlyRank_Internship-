## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*
Data Overview: We load the starter dataset to check total pages, average impressions, and the percentage of declining pages to confirm the problem size is substantial.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import os
import pandas as pd
import urllib.request

file_path = "content_refresh_anonymized.csv"
data_url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"

if not os.path.exists(file_path):
    urllib.request.urlretrieve(data_url, file_path)

df = pd.read_csv(file_path)
total_pages = len(df)
declining_rate = (df["trend_direction"].str.lower() == "down").mean() * 100

print(f"Total Pages: {total_pages:,}")
print(f"Declining Pages Rate: {declining_rate:.1f}%")
print(f"Average Impressions: {df['impressions_90d'].mean():.1f}")
```
