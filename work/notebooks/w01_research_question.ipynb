## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os
import pandas as pd
import urllib.request

# Download the starter data directly from GitHub
file_path = "content_refresh_anonymized.csv"
data_url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"

if not os.path.exists(file_path):
    print("Downloading dataset...")
    urllib.request.urlretrieve(data_url, file_path)

# Load the dataset
df = pd.read_csv(file_path)

# Calculate numbers
total_pages = len(df)
decay_pct = (df["trend_direction"] == "down").mean() * 100
corr = df["search_volume"].corr(df["impressions_90d"])

print(f"1. Total pages in dataset: {total_pages:,}")
print(f"2. Percentage of decaying pages: {decay_pct:.1f}%")
print(f"3. Search volume vs impressions correlation: {corr:.3f}")
```
